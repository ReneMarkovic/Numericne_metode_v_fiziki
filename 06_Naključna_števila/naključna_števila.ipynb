$$X_{n+1} = (a * X_{n} + c)\mod m $$

```python
def LCG(X0:int,a:int,c:int,m:int):
    '''
    X0: seme
    a: množitelj (angl. multiplier),
    c: inkrement (angl. increment),
    m: modulo (angl. modulus).
    '''
    return (a*X0+c)%m

def generator(X0,a,c,m,N=1):
    if N==1:
        return LCG(X0,a,c,m)/m
    else:
        sample=[]
        sample.append(X0/m)
        for k in range(N-1):
            X1=LCG(X0,a,c,m)
            X0=X1
            sample.append(X0/m)
        return sample
    
def plot(sample):
    plt.figure(figsize=(21,6))
    plt.subplot(1,3,1)
    plt.plot(sample,'o-')
    plt.xlabel('$n$')
    plt.ylabel('$X_n$')
    
    plt.subplot(1,3,2)
    x=sample[0:-1]
    y=sample[1:]
    plt.scatter(x,y)
    plt.xlabel('$X_n$')
    plt.ylabel('$X_{n+1}$')
    
    plt.subplot(1,3,3)
    plt.hist(sample,bins=20)
    
    plt.show()
```

# Primer 1

```python
a=c=x0=5
m=10

sample=generator(x0,a,c,m,N=10)
print("Sample =",sample)
plot(sample)
```

# Hit or miss $\pi$

```python
import numpy as np
import matplotlib.pyplot as plt
N = 10_000
x = np.random.rand(N)
y = np.random.rand(N)
R = np.sqrt(x**2+y**2)
xs = np.array(sorted(x))
ys = np.sqrt(1-xs**2)

color = ['red' if R[i] > 1 else 'green' for i in range(N)]

plt.plot(xs,ys,color='black')
plt.scatter(x,y,color=color,s = 10/np.log10(N))
plt.xlabel('$x$')
plt.ylabel('$y$')
plt.title('Naključna števila v kvadratu')
plt.xlim(0,1)
plt.ylim(0,1)

result = 4*len([i for i in range(N) if R[i] <= 1])/N
print("Ocena za pi =",result)
```
